```python
x = np.linspace(-1,1,100)
a, b, c = 1, 2, 3
y_exact = a + b * x + c * x**2

m = 100
X = 1 - 2 * np.random.rand(m)
Y = a + b * X + c * X**2 + np.random.rand(m)
```

```python
A = np.vstack([X**0, X**1, X**2])
```

```python
sol, r, rank, sv = la.lstsq(A.T, Y)
sol
```

```python
y_fit = sol[0] + sol[1] * x + sol[2] * x**2
```

```python
fig, ax = plt.subplots(figsize=(12,4))
ax.plot(X, Y, 'go', alpha=0.5, label='Simulate data')
ax.plot(x, y_exact, 'k', lw=2, label='True value $y = 1 + 2x + 3x^2$')
ax.plot(x, y_fit,   'b', lw=2, label='Least square fit')
ax.set_xlabel(r'$x$', fontsize=18)
ax.set_ylabel(r'$y$', fontsize=18)
ax.legend(loc=2)
```

```python
A = sympy.Matrix(A)
A = A.transpose()

Y = sympy.Matrix(Y)
```

```python
A.solve_least_squares(Y)
```

```python
Y = Y = a + b * X + c * X**2 + np.random.rand(m)
```

```python
A = np.vstack([X**n for n in range(2)])
sol, r, rank, sv = la.lstsq(A.T, Y)
y_fit1 = sum([s * x**n for n, s in enumerate(sol)])
```

```python
A = np.vstack([X**n for n in range(16)])
sol, r, rank, sv = la.lstsq(A.T, Y)
y_fit15 = sum([s * x**n for n, s in enumerate(sol)])
```

```python
fig, ax = plt.subplots(figsize=(12,4))
ax.plot(X, Y, 'go', alpha=0.5, label='Simulate data')
ax.plot(x, y_exact, 'k', lw=2, label='True value $y = 1 + 2x + 3x^2$')
ax.plot(x, y_fit1,  'b', lw=2, label='Least square fit [ 1st order]')
ax.plot(x, y_fit15, 'm', lw=2, label='Least square fit [15st order]')
ax.set_xlabel(r'$x$', fontsize=18)
ax.set_ylabel(r'$y$', fontsize=18)
ax.legend(loc=2)
plt.tight_layout()
```
